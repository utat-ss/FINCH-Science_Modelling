```python
#Selecting the perfect soil samples

data_list = ['NF_068', 'NF_083', 'QUE_T1.05.A', 'QUE_T1.05.C', 'QUE_T1.05.E', 
             'QUE_T1.06.B', 'QUE_T2.06.B', 'QUE_T2.06.F', 'QUE_T2.06.I', 
             'QUE_T3.51.B', 'QUE_T3.06']

print(data[0])
"""
    - col1: Name
    - col5: RWC index
    - col8: 400 nm
    - col217: 2490 nm
"""
samples = {}
for record in data:
    if record['col1'] in data_list:
        samples[record['col1']] = record
```

```python
# If XML

import xml.etree.ElementTree as ET
tree = ET.parse('test_xml.xml')
root = tree.getroot()
print(root.tag)

for child in root:
    print(child.tag, child.attrib)

for rank in root.iter('rank'):
    new_rank = int(rank.text) + 1
    rank.text = str(new_rank)
    rank.set('updated', 'yes')

tree.write('output.xml')
```

```python
tree = ET.parse('coeff_diff.xml')
root = tree.getroot()
```

```python
for data_ in samples:
    index = 8
    for wavelength in root.iter('lambertianMultiplicativeFactorForLUT'):
        new_reflect = samples[data_][f'col{index}']
        wavelength.set('reflectanceFactor', str(new_reflect))
        new_diff_transmit = 0
        wavelength.set('diffuseTransmittanceFactor', str(new_diff_transmit))
        new_dir_transmit = 0
        wavelength.set('directTransmittanceFactor', str(new_dir_transmit))
        new_spec_intensity = 0.05
        wavelength.set('specularIntensityFactor', str(new_spec_intensity))
        index += 1
        
    
    tree.write(f'coeff_diff-{data_}.xml')

        # print(new_reflect)
```
